## 1. Introduction to pandas

Pandas is the main library for working with data in Python. Its core structure is the DataFrame:

```python
import pandas as pd
import numpy as np

# Create a simple DataFrame from a dictionary
weather_data = {
    'date': ['2024-01-01', '2024-01-02', '2024-01-03'],
    'temperature': [10, 12, 8],
    'precipitation': [0.5, 2.3, 0.0],
    'location': ['Germany', 'Germany', 'Germany']
}

df = pd.DataFrame(weather_data)
print(df)
print(f"\nDataFrame shape: {df.shape}")  # (rows, columns)
```

## 2. Loading Data from CSV Files

Your project has CSV data files. They are stored in the `data` folder of the course repository on GitHub, so pandas can load them directly from the web — this works the same in Google Colab and on your laptop.

We store the location of the data folder once in the variable `DATA_URL` and then just add the file name. (Working offline with the repository cloned to your laptop? Change it to `DATA_URL = '../data/'`.)

```python
# Where the data files are stored (use '../data/' instead if you work offline)
DATA_URL = 'https://raw.githubusercontent.com/Motzemoere/Earth-System-Modelling/main/data/'

# Load the Germany dataset
df_germany = pd.read_csv(DATA_URL + 'Data_swbm_Germany.csv')

# Look at the first few rows
print("First 5 rows:")
print(df_germany.head())

# Get basic info
print("\nDataFrame info:")
print(df_germany.info())
```

## 3. Exploring Your Data

```python
# Get statistics
print(df_germany.describe())

# Get column names
print(f"\nColumns: {df_germany.columns.tolist()}")

# Get data types
print(f"\nData types:\n{df_germany.dtypes}")
```

## 4. Accessing Data

There are several ways to access data in a DataFrame:

```python
# Access a single column
precip = df_germany['tp_[mm]']
print(f"Precipitation column (first 5 values):\n{precip.head()}")

# Basic statistics on a column
print(f"\nTotal precipitation: {precip.sum():.1f} mm")
print(f"Mean precipitation: {precip.mean():.2f} mm")
print(f"Max precipitation: {precip.max():.2f} mm")
print(f"Min precipitation: {precip.min():.2f} mm")

# Access by row index
first_row = df_germany.iloc[0]
print(f"\nFirst row:\n{first_row}")
```

## 5. Filtering Data

```python
# Find days with high precipitation
high_precip = df_germany[df_germany['tp_[mm]'] > 10]
print(f"Days with precipitation > 10mm: {len(high_precip)} days")
print(high_precip[['time', 'tp_[mm]']].head())

# Find days with low soil moisture
low_sm = df_germany[df_germany['sm_[m3/m3]'] < 0.2]
print(f"\nDays with soil moisture < 0.2: {len(low_sm)} days")
```

## 6. Introduction to NumPy Arrays

NumPy is for fast numerical computations. It works with arrays:

```python
# Create arrays
temps = np.array([15, 16, 18, 22, 25, 28, 29])
precips = np.array([10, 15, 5, 20, 30, 25, 18])

print(f"Temperatures: {temps}")
print(f"Precipitation: {precips}")

# Element-wise operations
rainfall_to_precip_ratio = precips / 2
print(f"\nRatio (precip/2): {rainfall_to_precip_ratio}")

# Mathematical operations
evaporation = np.sin(temps / 10) * 5  # Simplified example
print(f"\nEvaporation (example): {evaporation.round(2)}")

# Common operations
print(f"\nSum of precipitation: {precips.sum()}")
print(f"Mean temperature: {temps.mean():.1f}")
print(f"Max temperature: {temps.max()}")
```

## 7. Converting Between Pandas and NumPy

You often need to convert between DataFrames and arrays:

```python
# Create a small DataFrame
small_df = pd.DataFrame({
    'precip': [10, 15, 20],
    'temp': [15, 18, 22]
})

# Convert column to NumPy array
precip_array = np.asarray(small_df['precip'])
print(f"Type: {type(precip_array)}")
print(f"Array: {precip_array}")

# Now perform fast NumPy operations
result = precip_array * 2
print(f"\nAfter multiplying by 2: {result}")
```

## 8. Comparing Parts of a Dataset

The `time` column holds dates as text in the format `YYYY-MM-DD`. Text in this format sorts like dates, so you can filter time periods with `>=` and `<=`, just like numbers. Here we compare 2017 with 2018, a year with a severe summer drought in central Europe:

```python
# Select two years from the Germany dataset
df_2017 = df_germany[(df_germany['time'] >= '2017-01-01') & (df_germany['time'] <= '2017-12-31')]
df_2018 = df_germany[(df_germany['time'] >= '2018-01-01') & (df_germany['time'] <= '2018-12-31')]

print(f"2017 shape: {df_2017.shape}")
print(f"2018 shape: {df_2018.shape}")

# Compare average precipitation and soil moisture
print(f"\nAverage precipitation:")
print(f"2017: {df_2017['tp_[mm]'].mean():.2f} mm")
print(f"2018: {df_2018['tp_[mm]'].mean():.2f} mm")
print(f"\nAverage soil moisture:")
print(f"2017: {df_2017['sm_[m3/m3]'].mean():.3f} m3/m3")
print(f"2018: {df_2018['sm_[m3/m3]'].mean():.3f} m3/m3")
```

## Exercises

### Exercise 1: Explore the Germany Dataset
Load the Germany data and answer these questions:
- How many days of data are there?
- What is the mean soil moisture?
- What is the maximum runoff value?

```python
# YOUR CODE HERE
# df = pd.read_csv(DATA_URL + 'Data_swbm_Germany.csv')
# num_days = ...
# mean_sm = ...
# max_runoff = ...
df = pd.read_csv(DATA_URL + 'Data_swbm_Germany.csv')
num_days = len(df['time'])
mean_sm = df['sm_[m3/m3]'].mean()
max_runoff = df['ro_[m]'].max()

print(
    f"Number of days: {num_days},\n"
    f"Mean soil moisture: {mean_sm:.2f} m3/m3,\n"
    f"Maximum runoff: {max_runoff:.5f} m"
)
```

### Exercise 2: Filter and Compare
Split the Germany data into two periods (2008–2012 and 2013–2018) and compare:
- Which period has higher average precipitation?
- How many days had soil moisture > 0.4 in each period?

```python
# YOUR CODE HERE
df_early = df[(df['time'] >= '2008-01-01') & (df['time'] <= '2012-12-31')]
df_late = df[(df['time'] >= '2013-01-01') & (df['time'] <= '2018-12-31')]
avg_precip_early = df_early['tp_[mm]'].mean()
avg_precip_late = df_late['tp_[mm]'].mean()
high_sm_early = df_early[(df_early['sm_[m3/m3]'] >= 0.4)]
high_sm_late = df_late[(df_late['sm_[m3/m3]'] >= 0.4)]

print(f"Average precipitation (2008-2012): {avg_precip_early:.2f} mm, \nAverage precipitation (2008-2012): {avg_precip_late:.2f} mm, \n{len(high_sm_early)}, \n{len(high_sm_late)}")
```

### Exercise 3: NumPy Operations
Create NumPy arrays from precipitation and runoff data for Germany, then calculate:
- The proportion of rainfall that runs off (runoff / precipitation)
- The infiltration (precipitation - runoff)

```python
# YOUR CODE HERE
precip = np.asarray(df['tp_[mm]'])
runoff = np.asarray(df['ro_[m]'])
runoff_fraction = runoff.sum() / precip.sum()
infiltration = precip.sum() - runoff.sum()

print(f"Runoff (prop): {runoff_fraction:.2f}") 
print(f"Infiltration: {infiltration:.2f}")
```
