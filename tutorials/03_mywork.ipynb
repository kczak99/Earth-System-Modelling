## 1. For Loops - Repeating Operations

```python
# Simple loop over a list
temperatures = [15, 18, 22, 25, 20]

for temp in temperatures:
    print(f"Temperature: {temp}°C")

print("\n" + "="*40)

# Loop with range
print("Days 1 to 5:")
for day in range(1, 6):
    print(f"Day {day}")
```

## 2. Accumulating Values in Loops

A common pattern: calculate something step by step:

```python
# Calculate total rainfall
daily_precip = [5, 10, 2, 15, 8]
total_precip = 0

for precip in daily_precip:
    total_precip = total_precip + precip
    print(f"Added {precip}mm, total is now {total_precip}mm")

print(f"\nFinal total: {total_precip}mm")

# Same thing with multiplication
print("\n" + "="*40)
print("Calculating product:")
product = 1
numbers = [2, 3, 4]

for num in numbers:
    product = product * num
    print(f"Multiplied by {num}, product is now {product}")

print(f"Final product: {product}")
```

## 3. Loops with Conditions

Often you only want to do something in certain situations:

```python
# Count days with high rainfall
daily_precip = [5, 15, 2, 20, 8, 18, 3]
high_precip_days = 0

for precip in daily_precip:
    if precip > 10:
        print(f"High rainfall: {precip}mm")
        high_precip_days = high_precip_days + 1
    else:
        print(f"Normal rainfall: {precip}mm")

print(f"\nTotal days with high rainfall: {high_precip_days}")
```

## 4. While Loops - Repeat Until a Condition is Met

While loops are useful when you don't know how many iterations you need:

```python
# Example: Rain accumulation until we reach 50mm
import numpy as np

rainfall_data = [5, 8, 10, 12, 15, 6, 3]
total = 0
day = 0

while total < 50 and day < len(rainfall_data):
    total = total + rainfall_data[day]
    day = day + 1
    print(f"Day {day}: Total rainfall = {total}mm")

print(f"\nReached 50mm after {day} days")
```

## 6. Simulating Model Steps with Loops

Now let's combine loops and functions to simulate a simple water balance model:

```python
import numpy as np

# Simple simulation with NumPy arrays
num_days = 100

# Initialize arrays
temperature = np.zeros(num_days)
air_temperature = np.zeros(num_days)

# Set initial condition
temperature[0] = 5  # Cold starting temperature

# Simulate seasonal air temperature (simple sine curve)
for i in range(num_days):
    # Air temperature varies from 5°C to 25°C over the 100 days
    air_temperature[i] = 15 + 10 * np.sin(i / 100 * 2 * np.pi)

# Simulate water temperature
cooling_rate = 0.1

for i in range(num_days - 1):
    temp_diff = air_temperature[i] - temperature[i]
    temperature[i + 1] = temperature[i] + temp_diff * cooling_rate

# Summarize results
print("100-day water temperature simulation:")
print(f"Starting temperature: {temperature[0]:.1f}°C")
print(f"Ending temperature: {temperature[-1]:.1f}°C")
print(f"Mean water temperature: {temperature.mean():.1f}°C")
print(f"Max water temperature: {temperature.max():.1f}°C")
print(f"Min water temperature: {temperature.min():.1f}°C")
```

## 7. Storing Results in Lists and Arrays

### Exercise 2: Function with Multiple Parameters
Write a function that calculates the water balance change given current soil moisture, precipitation, and parameters. Then call it for different soil moisture values.

```python
def calculate_et(temperature, soil_moisture, max_soil_moisture):
    moisture_fraction = soil_moisture / max_soil_moisture
    et = temperature * moisture_fraction
    return et

for moist in [10,20,30,40,50]:
    et = calculate_et(25, moist, 200)
    print(f"Soil Moisture: {moist}, ET: {et}")

```
